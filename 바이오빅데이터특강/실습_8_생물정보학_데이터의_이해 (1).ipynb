```python
with open("example.fastq", "rt") as file:
    for record in SeqIO.parse(file, "fastq"):
        print(f"ID: {record.id}")
        print(f"Sequence: {record.seq}")
        print(f"Quality: {record.letter_annotations['phred_quality']}")
        quality = record.letter_annotations['phred_quality']
        quality_str = ''.join(chr(q + 33) for q in quality)
        print(f"Quality: {quality_str}\n")
        break
```

```python
# 1. 전체 record 숫자 구하기
# 2. 각 read(record)에 대한 평균 phread score값 구하기
# 3. 평균 30이 넘는 read와 넘지 않는 read숫자 세기
```

```python
import numpy as np
count = 0
scores = []
with open("example.fastq", "rt") as file:
    for record in SeqIO.parse(file, "fastq"):
        scores.append(np.array(record.letter_annotations['phred_quality']).mean())
        count += 1
print(count)
```

```python
scores = np.array(scores)
sum(scores>=2)
```
